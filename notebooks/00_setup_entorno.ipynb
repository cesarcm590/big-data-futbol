Objetivo: confirmar que el entorno conda `futbol-bigdata` tiene todas las

y selecciona el kernel `futbol-bigdata` en este notebook.

````python
import importlib

paquetes = [
    "pandas", "numpy", "matplotlib", "seaborn",
    "mplsoccer", "soccerdata", "statsbombpy",
    "requests", "bs4", "lxml", "highlight_text",
]

faltantes = []
for paquete in paquetes:
    try:
        importlib.import_module(paquete)
        print(f"OK  - {paquete}")
    except ImportError:
        faltantes.append(paquete)
        print(f"FALTA - {paquete}")

if faltantes:
    print("\nInstala lo que falte con:")
    print(f"pip install {' '.join(faltantes)}")
else:
    print("\nTodo listo.")
````
